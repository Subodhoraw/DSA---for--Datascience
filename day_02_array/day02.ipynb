```python
class legoBrick:
    def __init__(self,value):
        self.data = value #Put a number on the brick 
        self.next = None #its not snapped it into anything yet

# lets create three bricks 
brick1 = legoBrick(10)
brick2 = legoBrick(20)
brick3 = legoBrick(30)

#3.snap them together 
brick1.next = brick2 #Brick
brick2.next = brick3 #brick2 points to brick3 

#how to walk through the chain 
current_hand = brick1 #start the first brick

while current_hand is not None:
    print(f" I am holding brick with value :{current_hand.data}")

    current_hand = current_hand.next #move youre hand to next brick
```
